# Bab 2: Fundamental Concepts — How Do Machines Learn?

Notebook ini merangkum Bab 2 buku *Grokking Deep Learning* karya Andrew W. Trask. Bab ini masih bersifat konseptual (belum ada kode), membahas kosakata dan konsep dasar yang akan dipakai sepanjang buku: apa itu deep learning, machine learning, dan artificial intelligence, serta dua cara membagi algoritma machine learning — supervised vs unsupervised, dan parametric vs nonparametric.

## 1. Deep Learning, Machine Learning, dan Artificial Intelligence

Deep learning adalah subset dari machine learning, yang merupakan bidang yang didedikasikan untuk mempelajari dan mengembangkan mesin yang bisa belajar (terkadang dengan tujuan akhir mencapai general artificial intelligence). Hubungannya bertingkat:

**Artificial Intelligence ⊃ Machine Learning ⊃ Deep Learning**

Di industri, deep learning dipakai untuk menyelesaikan tugas-tugas praktis di berbagai bidang seperti computer vision (gambar), natural language processing (teks), dan automatic speech recognition (audio). Singkatnya, deep learning adalah subset metode dalam toolbox machine learning yang utamanya memakai artificial neural network, sejenis algoritma yang terinspirasi dari otak manusia.

Penting dicatat: tidak semua deep learning ditujukan untuk mengejar general artificial intelligence (mesin yang sadar diri seperti di film-film). Banyak aplikasinya justru dipakai untuk menyelesaikan beragam masalah praktis di industri.

## 2. Apa Itu Machine Learning?

> "A field of study that gives computers the ability to learn without being explicitly programmed." — Atribusi untuk Arthur Samuel

Machine learning adalah subbidang ilmu komputer di mana mesin belajar menjalankan tugas yang tidak secara eksplisit diprogramkan. Singkatnya, mesin mengamati suatu pola dan mencoba menirunya, baik secara langsung maupun tidak langsung. Dua kategori utamanya:

- **Supervised learning**: peniruan langsung dari pola antara dua dataset. Selalu berusaha mengubah satu dataset input menjadi dataset output.
- **Unsupervised learning**: peniruan tidak langsung, karena tidak ada "jawaban benar" yang diketahui sebelumnya.

Beberapa contoh tugas supervised learning:

- Memakai piksel gambar untuk mendeteksi ada/tidaknya kucing
- Memakai riwayat film yang disukai untuk memprediksi film lain yang mungkin disukai
- Memakai kata-kata seseorang untuk memprediksi apakah dia senang atau sedih
- Memakai data sensor cuaca untuk memprediksi probabilitas hujan
- Memakai data berita untuk memprediksi harga saham besok

## 3. Supervised Machine Learning

Supervised learning adalah metode untuk mengubah satu dataset menjadi dataset lain. Contohnya, kalau ada dataset "Harga Saham Senin" selama 10 tahun terakhir dan dataset "Harga Saham Selasa" di periode yang sama, algoritma supervised learning akan mencoba memakai salah satu untuk memprediksi yang lain.

Secara skematis:

```
Apa yang kita tahu  ->  [Supervised Learning]  ->  Apa yang ingin kita tahu
```

Supervised machine learning adalah "roti dan mentega"-nya applied artificial intelligence (juga dikenal sebagai narrow AI). Bermanfaat untuk mengubah apa yang kita ketahui sebagai input menjadi apa yang ingin kita ketahui, dengan cepat. Bahkan unsupervised learning biasanya dilakukan untuk membantu mengembangkan supervised learning yang lebih akurat.

## 4. Unsupervised Machine Learning

Unsupervised learning juga mengubah satu dataset jadi dataset lain, tapi dataset hasil transformasinya tidak diketahui atau dipahami sebelumnya. Tidak ada "jawaban benar" yang ingin ditiru model; kita hanya meminta algoritma untuk "temukan pola dalam data ini dan ceritakan padaku".

Contoh klasiknya adalah **clustering**: mengelompokkan dataset menjadi beberapa grup. Clustering mengubah serangkaian titik data menjadi serangkaian label cluster (biasanya berupa angka, karena algoritmanya sendiri tidak tahu arti atau nama dari cluster yang ditemukan).

Contoh ilustratif dari buku: kata-kata seperti "puppies", "kittens" dikelompokkan jadi cluster 1, sementara "pizza", "hot dog", "burger" dikelompokkan jadi cluster 2. Algoritmanya tidak memberi tahu bahwa cluster 1 berarti "lucu" dan cluster 2 berarti "lezat", kita sendiri yang menafsirkannya.

Poin penting: walau ada banyak bentuk unsupervised learning, semuanya bisa dipandang sebagai bentuk clustering.

## 5. Parametric vs. Nonparametric Learning

Selain dibagi jadi supervised/unsupervised, algoritma machine learning juga bisa dibagi jadi parametric dan nonparametric. Jadi total ada empat kombinasi: supervised-parametric, supervised-nonparametric, unsupervised-parametric, dan unsupervised-nonparametric.

- **Model parametric**: punya jumlah parameter yang tetap (fixed).
- **Model nonparametric**: jumlah parameternya tidak tetap, ditentukan oleh data (secara teori bisa tak terbatas).

Analogi dari buku: memasukkan pasak persegi ke lubang yang cocok. Sebagian (seperti bayi) langsung mencoba memasukkannya ke semua lubang sampai cocok (parametric, trial-and-error). Seorang remaja mungkin menghitung jumlah sisinya (empat) lalu mencari lubang dengan jumlah sisi yang sama (nonparametric, berbasis counting).

## 6. Supervised Parametric Learning

Secara oversimplified: trial-and-error learning memakai "knob" (kenop). Model supervised parametric learning punya jumlah knob yang tetap, dan belajar dengan cara memutar knob-knob itu. Tiga langkahnya:

1. **Predict**: data masuk, diproses berdasarkan posisi knob, menghasilkan prediksi.
2. **Compare**: prediksi dibandingkan dengan truth pattern (jawaban sebenarnya).
3. **Learn**: knob disesuaikan berdasarkan seberapa jauh prediksi meleset dan data input saat prediksi dibuat, supaya prediksi berikutnya lebih akurat.

Contoh dari buku: memprediksi probabilitas tim Red Sox menang di World Series berdasarkan statistik olahraga. Kalau prediksinya 98% tapi ternyata tim itu kalah, model akan menyesuaikan knob (misalnya mengecilkan pengaruh knob "jumlah fans" jika itu bukan prediktor yang bagus) agar prediksi berikutnya lebih akurat untuk data serupa.

Seluruh "pengetahuan" yang dipelajari model tersimpan dalam posisi knob-knobnya pada waktu tertentu. Ini mirip algoritma pencarian: mencari konfigurasi knob yang optimal lewat coba-coba dan penyesuaian berulang.

## 7. Unsupervised Parametric Learning

Pendekatannya serupa, tapi dipakai untuk mengelompokkan data. Biasanya ada beberapa knob untuk tiap grup, masing-masing memetakan afinitas data input terhadap grup tertentu.

Contoh dari buku: mengelompokkan data pertandingan (kandang/tandang, jumlah penonton) ke dalam tiga grup. Tiap grup punya "mesin" sendiri yang mengubah data input jadi angka antara 0 dan 1, menunjukkan probabilitas data itu menjadi anggota grup tersebut. Misalnya, satu titik data bisa punya probabilitas 94% milik grup 1, 1% grup 2, dan 5% grup 3.

## 8. Nonparametric Learning

Secara oversimplified: metode berbasis counting (menghitung). Jumlah parameter pada model nonparametric berubah berdasarkan data, bukan ditentukan sebelumnya.

Contoh dari buku: model yang menghitung berapa kali warna lampu lalu lintas tertentu membuat mobil "jalan". Setelah menghitung beberapa contoh, model bisa menyimpulkan lampu tengah selalu (100%) membuat mobil jalan, sementara lampu kanan hanya kadang (50%). Kalau ada lima lampu, model punya lima parameter (lima hitungan), karena jumlah parameternya memang mengikuti jumlah hal yang dihitung dalam data.

Penting: istilah "parameter" sendiri bersifat generik, merujuk pada kumpulan angka yang dipakai untuk memodelkan suatu pola, apa pun bentuknya (hitungan, bobot/weight, koefisien korelasi, dan seterusnya). Deep learning sendiri termasuk kelas model parametric, dan buku ini tidak akan membahas model nonparametric lebih jauh.

## Poin-poin Penting Bab 2

- Deep learning adalah subset dari machine learning, yang merupakan subset dari artificial intelligence.
- Machine learning dibagi menjadi supervised learning (meniru pola langsung antara dua dataset) dan unsupervised learning (mengelompokkan data tanpa jawaban benar yang diketahui sebelumnya, pada intinya berupa clustering).
- Machine learning juga dibagi menjadi parametric (jumlah parameter tetap, belajar lewat penyesuaian "knob") dan nonparametric (jumlah parameter mengikuti data, umumnya berbasis counting).
- Deep learning memakai neural network untuk melakukan prediksi, baik secara supervised maupun unsupervised, dan termasuk kelas model parametric.
- Bab berikutnya mulai bersifat proyek: neural network pertama akan dibangun langsung.

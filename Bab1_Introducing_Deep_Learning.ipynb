## 1. Mengapa Deep Learning Penting Dipelajari

Deep learning adalah alat yang ampuh untuk mengotomasi kecerdasan secara bertahap. Sejak awal peradaban, manusia terus membangun alat yang makin baik untuk memahami dan mengendalikan lingkungan sekitarnya, dan deep learning adalah babak terbaru dari cerita inovasi ini.

Yang membuat bidang ini istimewa adalah sifatnya yang lebih dekat ke inovasi mental ketimbang mekanis. Deep learning memakai algoritma yang terinspirasi dari otak manusia (neural network/jaringan saraf tiruan) dan algoritma yang relatif sama ini terbukti berhasil di berbagai bidang berbeda, mulai dari computer vision, pengenalan suara, sampai penerjemahan bahasa.

Buku ini juga menyinggung dampak deep learning terhadap otomasi pekerjaan terampil, dari operator call center sampai sopir taksi dan analis bisnis tingkat bawah. Penulis menekankan bahwa dampak ini sudah terasa bahkan jika perkembangan teknologi berhenti hari ini juga, karena kekuatan teknologi yang sudah ada saat ini.
